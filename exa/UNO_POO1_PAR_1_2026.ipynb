1. Una empresa de transporte desea registrar información de sus vehículos.
* Crear una clase **Vehiculo** (que no se pueda instanciar) con los siguientes atributos: **marca**, **modelo** (default: '') y **consumo_100km** (consumo de combustible en litros cada 100 km. default: 0). Incluir un método **info()** que muestre los datos del vehículo
* Crear una clase **Auto** que herede de Vehiculo y agregue: **cantidad_puertas** (default: 0)
* Crear una clase **Camion** que herede de Vehiculo y agregue: **capacidad_carga** (default: 0)
* Sobrescribir el método info() en ambas clases para mostrar todos los datos
* Implementar el método **calcular_consumo(cant_km)**
* Crear un objeto de cada tipo y mostrar su información.

```python
from abc import abstractmethod, ABC

class Vehiculo(ABC):

    def __init__(self, marca, modelo='', consumo_100km=0):
        self.__marca = marca
        self.__modelo = modelo
        self.__consumo_100km = consumo_100km

    @abstractmethod
    def info(self):
        return f'{self.__marca}, {self.__modelo}, {str(self.__consumo_100km)}'

    def calcular_consumo(self, cant_km):
        return self.__consumo_100km * cant_km / 100

class Auto(Vehiculo):

    def __init__(self, marca, modelo='', consumo_100km=0, cant_puertas=0):
        super().__init__(marca, modelo, consumo_100km)
        self.__cant_puertas = cant_puertas

    def info(self):
        return f'{super().info()}, {str(self.__cant_puertas)}'

class Camion(Vehiculo):

    def __init__(self, marca, modelo='', consumo_100km=0, cap_carga=0):
        super().__init__(marca, modelo, consumo_100km)
        self.__cap_carga = cap_carga

    def info(self):
        return f'{super().info()}, {str(self.__cap_carga)}'
        
auto = Auto('Toyota', 'Corolla', 4, 8)
camion = Camion('Mercedes-Benz', 'Actros', 20, 30)
print(auto.info())
print(camion.info()) 
print(auto.calcular_consumo(10))
print(camion.calcular_consumo(10))
```

2. Implementar una función **calcular_costo_combustible(lista_vehiculos, km, valor_litro=2100.0)**. Realizar un test que verifique que el método funcione correctamente.

```python
def calcular_costo_combustible(lista_vehiculos, km, valor_litro=2100.0):
    acum = 0
    for vehiculo in lista_vehiculos:
        acum += vehiculo.calcular_consumo(km) * valor_litro
    return acum    
```

```python
import unittest

class TestCombustible(unittest.TestCase):

    def test_costo_combustible(self):
        lista = [auto, camion]
        self.assertEqual(5040.0, calcular_costo_combustible(lista, 10))

unittest.main(defaultTest='TestCombustible', argv=[''], verbosity=2, exit=False)
```
